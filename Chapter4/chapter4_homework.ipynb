```python
import os
import openpyxl as ol
import pandas as pd
```

# Nhập Xuất Kho

```python
file_name = "data_nhap_xuat_kho.xlsx"
if os.path.exists(file_name):
    wb = ol.load_workbook(file_name)
    ws = wb['Sheet1']
    max_row = ws.max_row
    
    max_collumn = ws.max_column
    for col in range(1, max_collumn + 1):
        if ws.cell(row=1, column=col).value == "Lợi nhuận":
            loi_nhuan_col = col
        if ws.cell(row=1, column=col).value == "Phân loại Lợi nhuận":
            loi_nhuan_phan_loai_col = col
             
    if loi_nhuan_col is None:
        ws.cell(row=1, column=ws.max_column + 1, value="Lợi nhuận")
    if loi_nhuan_phan_loai_col is None:
        ws.cell(row=1, column=ws.max_column + 1, value="Phân loại Lợi nhuận")
        
    for row_num, row in enumerate(ws.iter_rows(min_row=2, max_row=max_row, values_only=True), start=2):
        phan_loai = ""
        ma_san_pham = row[0]
        ten_san_pham = row[1]
        ngay_xuat = row[2]
        so_luong = int(row[3])
        gia_ban = float(row[4])
        gia_nhap = float(row[5])
        loi_nhuan = (gia_ban - gia_nhap) * so_luong
        
        if loi_nhuan > 200000:
            phan_loai = "Lời nhiều"
        elif 100000 < loi_nhuan <= 200000:
            phan_loai = "Lời trung bình"
        elif loi_nhuan < 100000:
            phan_loai = "Lời ít"
        else:
            phan_loai = "Lỗ"
        
        ws.cell(row=row_num, column=loi_nhuan_col, value=str(loi_nhuan))
        ws.cell(row=row_num, column=loi_nhuan_phan_loai_col, value=phan_loai)
        
    wb.save(file_name)
    
```

# Logistic

```python
logistic_file_name = "data_logistics.xlsx"
sheet_name = "Dữ liệu"

if os.path.exists(logistic_file_name):
    wb = ol.load_workbook(logistic_file_name)
    ws = wb[sheet_name]
    max_row = ws.max_row
    phan_loai_cuoc_col = None
    phi_van_chuyen_col = None
    
    max_collumn = ws.max_column
    for col in range(1, max_collumn + 1):
        if ws.cell(row=1, column=col).value == "Phí vận chuyển (đ)":
            phi_van_chuyen_col = col
        if ws.cell(row=1, column=col).value == "Phân loại cước":
            phan_loai_cuoc_col = col
             
    if phan_loai_cuoc_col is None:
        ws.cell(row=1, column=ws.max_column + 1, value="Phân loại cước")
    if phi_van_chuyen_col is None:
        ws.cell(row=1, column=ws.max_column + 1, value="Phí vận chuyển (đ)")
        
    mien_bac = ['Hà Nội', 'Hải Phòng', 'Nam Định']
    mien_trung = ['Đà Nẵng', 'Huế', 'Nha Trang']
    mien_nam = ['TP.HCM', 'Vũng Tàu', 'Cần Thơ']
    
    def tim_mien(ten_tinh):
        if ten_tinh in mien_bac:
            return "Bắc"
        elif ten_tinh in mien_trung:
            return "Trung"
        elif ten_tinh in mien_nam:
            return "Nam"
        else: 
            return "Khác"
        
    for row_num, row in enumerate(ws.iter_rows(min_row=2, max_row=max_row, values_only=True), start=2):
        ma_don = row[0]
        khach_hang = row[1]
        loai_hang = row[2]
        noi_gui = row[3]
        noi_nhan = row[4]
        khoi_luong = float(row[5])
        loai_van_chuyen = row[6]
        cuoc_co_ban = float(row[7])
        phi_van_chuyen = cuoc_co_ban * khoi_luong
        loai_van_chuyen = ""
        giam_gia = 0
        giam_gia_mien = 0
        phan_loai_cuoc = ""
        
        if khoi_luong <= 5:
            giam_gia = 0
        elif 5 < khoi_luong <= 20:
            giam_gia = 0.05
        elif khoi_luong > 20:
            giam_gia = 0.1
        
        tong_tien = phi_van_chuyen * (1 - giam_gia)
        
        if loai_van_chuyen == "Siêu tốc":
            tong_tien += 20000
        
        mien_gui = tim_mien(noi_gui)
        mien_nhan = tim_mien(noi_nhan)
        
        if mien_gui == mien_nhan:
            giam_gia_mien = 0.05
            
        tong_tien = tong_tien * (1 - giam_gia_mien)
            
        if tong_tien > 500000:
            phan_loai_cuoc = "Cao"
        elif 200000 < tong_tien <= 500000:
            phan_loai_cuoc = "Trung Bình"
        else:
            phan_loai_cuoc = "Thấp"
        
        ws.cell(row=row_num, column=phi_van_chuyen_col, value=tong_tien)
        ws.cell(row=row_num, column=phan_loai_cuoc_col, value=phan_loai_cuoc)
        
    wb.save(logistic_file_name)
```
