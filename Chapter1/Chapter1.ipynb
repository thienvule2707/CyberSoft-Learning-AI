```python
kpi = float(input("Nhập KPI: "))
min_result = float(input("Nhập kết quả tối thiểu: "))
result = float(input("Nhập doanh số thực tế"))
base_salary = float(input("Nhập lương cơ bản"))

thu_nhap = 0

if min_result <= result >= kpi:
    thu_nhap = base_salary + base_salary * 0.2
    print(f"Lương: {thu_nhap}")
if min_result <= result <= kpi:
    thu_nhap = base_salary
    print(f"Lương: {thu_nhap}")
if min_result >= result:
    print("Warning")
```
